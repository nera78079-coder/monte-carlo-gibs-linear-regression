```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def gibbs_sampling(num_samples, burn_in=100):
    samples = []

    # initial values
    x, y = 0, 0

    for i in range(num_samples + burn_in):
        # sample x given y
        x = np.random.normal(loc=y, scale=1)

        # sample y given x
        y = np.random.normal(loc=x, scale=1)

        if i >= burn_in:
            samples.append([x, y])

    return np.array(samples)
```

```python
samples = gibbs_sampling(5000)
samples
```

```python
plt.scatter(samples[:, 0], samples[:, 1], alpha=0.3)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Gibbs Sampling")
plt.show()
```
